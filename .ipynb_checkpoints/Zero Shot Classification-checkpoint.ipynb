```python
from transformers import pipeline, logging
logging.set_verbosity_error() # view fewer warnings
```

```python
from transformers import pipeline

model = pipeline(
    "zero-shot-classification",
    model="MoritzLaurer/ModernBERT-base-zeroshot-v2.0",
    device=-1
)
```

```python
text = 'I didn\'t like the taste of that lemonade at all.'
categories = ['quote', 'food & drinks', 'technology']
result = model(text, candidate_labels = categories)
print(result)
```

```python
def get_category(sentence, categories):
    result = model(sentence, candidate_labels = categories)
    return result['labels'][0]
```

```python
text1 = 'When life gives you lemons, make lemonade! 🙂'
text2 = 'A dozen lemons will make a gallon of lemonade.'
text3 = 'I didn\'t like the taste of that lemonade at all.'
text4 = 'i am software developer. i love coding'

categories = ['quote', 'food & drinks', 'technology']
print(get_category(text1, categories))
print(get_category(text2, categories))
print(get_category(text3, categories))
print(get_category(text4, categories))
```
